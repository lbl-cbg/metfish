```python
import mdtraj as md
import os
import glob
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import pickle
import os
import shutil
from metfish.utils import get_Pr
```

```python
output_path='/global/cfs/cdirs/m4704/100125_Nature_Com_data/single_conformation/nmr'
openfold_path='/global/cfs/cdirs/m4704/100125_Nature_Com_data/single_conformation/af'
ref_path='/global/cfs/cdirs/m4704/100125_Nature_Com_data/Apo_holo_data/pdbs'
```

```python
plt.rcParams.update({'font.size': 16})
```

```python
def pr_compare(target_path, openfold_path,ref_path,pdb_id):
    ref_saxs=get_Pr(os.path.join(ref_path,pdb_id+'.pdb'), None, None, 0.5)
    target_saxs=get_Pr(os.path.join(target_path,pdb_id+'.pdb'), None, None, 0.5)
    openfold_saxs=get_Pr(os.path.join(openfold_path,pdb_id+'.pdb'), None, None, 0.5)
    
    r_ref, Pr_ref = ref_saxs
    r_target, Pr_target = target_saxs
    r_openfold, Pr_openfold = openfold_saxs
    
    plt.figure(figsize=(6,6))
    plt.plot(r_ref, Pr_ref, label='Ground Truth', lw=2)
    plt.plot(r_target, Pr_target, label='AlphaSAXS', lw=2)
    plt.plot(r_openfold, Pr_openfold, label='OpenFold', lw=2)
    plt.xlabel(r'$r$ ($\AA$)')
    plt.ylabel('$P(r)$')
    plt.title(f"P(r) Comparison for {pdb_id}")
    plt.legend()
    plt.tight_layout()
    plt.show()
```

```python
pr_compare(output_path, openfold_path,ref_path,'1EEJ_B')
```

```python
pr_compare(output_path, openfold_path,ref_path,'1FMF-4_A')
```
